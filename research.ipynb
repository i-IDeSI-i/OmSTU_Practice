# Тесты вычисления определённого интеграла (task14)

Ниже приведены тесты метода `DefiniteIntegral.Solve`. Тесты выполняются без модификаций.

Результат прогона через `dotnet test`: **всего 5, успешно 5, сбоев 0**.

```polyglot-notebook
#r "task14/bin/Release/net8.0/task14.dll"

using task14;
using System;

var X = (double x) => x;
var SIN = (double x) => Math.Sin(x);

// Тесты из задания task14 — выполняются без модификаций
Console.WriteLine(Math.Abs(DefiniteIntegral.Solve(-1, 1, X, 1e-4, 2) - 0) < 1e-4);
Console.WriteLine(Math.Abs(DefiniteIntegral.Solve(-1, 1, SIN, 1e-5, 8) - 0) < 1e-4);
Console.WriteLine(Math.Abs(DefiniteIntegral.Solve(0, 5, X, 1e-6, 8) - 12.5) < 1e-5);
```
